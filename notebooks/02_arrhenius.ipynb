```python
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("..")
from src.diffusion import arrhenius_diffusivity, D0_EXAMPLE, Q_EXAMPLE

T = np.linspace(280, 450, 100)   # kelvin
D = arrhenius_diffusivity(D0_EXAMPLE, Q_EXAMPLE, T)

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(1000 / T, np.log(D))
ax.set_xlabel("1000 / T  (1/K)")
ax.set_ylabel("ln D   (D in m²/s)")
ax.set_title("Arrhenius plot: hydrogen diffusivity in BCC iron")
ax.grid(alpha=0.3)
fig.tight_layout()
fig.savefig("../figures/arrhenius_plot.png", dpi=300)
plt.show()
```
